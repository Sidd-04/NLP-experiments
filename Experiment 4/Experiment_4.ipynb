## 4.1 Term Frequency and NER using toolkits

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
import nltk
import pandas as pd
from collections import Counter
import spacy

file_name = "4.1_4.2_input.txt"

with open(file_name, "r", encoding="utf-8") as f:
    text = f.read()

# Tokenization using NLTK
words = nltk.word_tokenize(text.lower())

# Keep only alphabetic words
words = [word for word in words if word.isalpha()]

# Calculate frequency
frequency = Counter(words)

# Save to CSV
df = pd.DataFrame(frequency.items(), columns=["Term", "Frequency"])
df.to_csv("term_frequency_toolkit.csv", index=False)

print("Top 10 Terms:")
print(df.sort_values("Frequency", ascending=False).head(10))

# NER using spaCy
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

## 4.2 Term Frequency without using any NLP toolkit

```python
import string
import pandas as pd

file_name = "4.1_4.2_input.txt"

with open(file_name, "r", encoding="utf-8") as f:
    text = f.read()

# Lowercase
text = text.lower()

# Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))

# Simple tokenization
words = text.split()

# Calculate frequency using dictionary
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

# Convert to DataFrame
df = pd.DataFrame(
    frequency.items(),
    columns=["Term", "Frequency"]
)

# Save CSV
df.to_csv("term_frequency_without_toolkit.csv", index=False)

# Top 10 terms
top10 = df.sort_values(
    "Frequency",
    ascending=False
).head(10)

print("Top 10 Terms:")
print(top10)
```

## 4.3 TF-IDF without using NLP toolkits

```python
import math
import string
from collections import Counter

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# Step 1: Preprocess documents
processed_docs = []

for text in documents:
    text = text.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    words = text.split()
    processed_docs.append(words)

print("Processed Documents:")
for i, doc in enumerate(processed_docs, 1):
    print(f"Document {i}:", doc)

# Step 2: Calculate TF
tf_list = []

for doc in processed_docs:
    tf = Counter(doc)
    tf_list.append(tf)

print("\nTF:")
for i, tf in enumerate(tf_list, 1):
    print(f"Document {i}:", dict(tf))

# Step 3: Calculate DF
df = Counter()

for doc in processed_docs:
    unique_words = set(doc)
    for word in unique_words:
        df[word] += 1

print("\nDF:")
print(dict(df))

# Step 4: Calculate IDF
N = len(documents)
idf = {}

for word, value in df.items():
    idf[word] = math.log(N / value)

print("\nIDF:")
for word, value in idf.items():
    print(word, ":", round(value, 4))

# Step 5: Calculate TF-IDF
print("\nTF-IDF and Top 10 Terms")

for i, tf in enumerate(tf_list, 1):
    scores = {}

    for word, value in tf.items():
        scores[word] = value * idf[word]

    print(f"\nDocument {i}:")
    print("Term\tTF\tDF\tIDF\tTF-IDF")

    for word in scores:
        print(
            word,
            "\t",
            tf[word],
            "\t",
            df[word],
            "\t",
            round(idf[word], 4),
            "\t",
            round(scores[word], 4)
        )

    top10 = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )[:10]

    print("\nTop 10 TF-IDF Terms:")
    for word, score in top10:
        print(word, "->", round(score, 4))
```
